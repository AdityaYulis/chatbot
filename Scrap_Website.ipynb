```python
import requests
from bs4 import BeautifulSoup
import json
import os
from datetime import datetime
```

```python
def scrape_page(url, category, filename):

    response = requests.get(url)
    response.raise_for_status()

    soup = BeautifulSoup(response.text, "html.parser")

    title = soup.title.text.strip() if soup.title else filename

    text = soup.get_text(
        separator="\n",
        strip=True
    )

    data = {
        "title": title,
        "url": url,
        "category": category,
        "scraped_at": datetime.now().isoformat(),
        "content": text
    }

    folder = f"data/{category}"
    os.makedirs(folder, exist_ok=True)

    filepath = f"{folder}/{filename}.json"

    with open(
        filepath,
        "w",
        encoding="utf-8"
    ) as f:
        json.dump(
            data,
            f,
            ensure_ascii=False,
            indent=4
        )

    print(f"Saved: {filepath}")
```

```python
pages = [
    {
        "url": "https://dikemas.com/about-us",
        "category": "company",
        "filename": "about-us"
    },
    {
        "url": "https://dikemas.com/jasa-kemas",
        "category": "services",
        "filename": "jasa-kemas"
    },
    {
        "url": "https://dikemas.com/mesin-pengemas-sachet",
        "category": "services",
        "filename": "mesin-kemasan"
    }
]
```

```python
for page in pages:
    scrape_page(
        page["url"],
        page["category"],
        page["filename"]
    )
```
